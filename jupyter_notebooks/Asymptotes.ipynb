```python
# Import libs
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
# Asymptotic pricing function

# a = max price
# b = start price
# s = supply
# k = determines rate of gradient change

sum_price = 0
a = 500
b = 1
k = 0
capacity = 200
s_es = [] # supply
ys = [] # spot price on the curve
prices = []

for i in range(200):
    s_es.append(i)

for s in s_es:
    y = b + (a / (capacity + k - s))
    ys.append(y)
    sum_price += y
    prices.append(sum_price)
    
print(ys)
print()
print(prices)
print()
print(sum_price)
    
plt.figure(figsize=(10, 4))
plt.plot(s_es, ys)
plt.xlabel('Supply')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```

```python
# Calculating price using definite integral of price curve function


# a = max price
# b = start price
# s = supply
# q = quantity requested
# k = determines rate of gradient change

price = 0
a = 500
b = 1
k = 0
capacity = 200
s = 1
qs = []
prices = []

# Use a loop to calculate prices for different requested quantities of tokens
for q in range(200-s):
    qs.append(q)
    # price for quantity q calculated once for given supply s
    price = b*q + a*math.log((capacity + k - s) / (capacity + k - (s + q)))
    prices.append(price)
    
print(prices)
    
plt.figure(figsize=(10, 4))
plt.plot(qs, prices)
plt.xlabel('Supply')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```

```python
# Approximation using log2 as openzeppelin math library doesn't implement natural log

# a = max price
# b = start price
# s = supply
# q = quantity requested
# k = determines rate of gradient change

price = 0
a = 500
b = 1
k = 0
capacity = 200
s = 1
qs = []
prices = []


for q in range(200-s):
    qs.append(q)
    #price = b*q + a*math.log2((capacity + k - s) / (capacity + k - (s + q)))
    price = b*q + a*math.log2((capacity + k - s) / (capacity + k - s - q))
    prices.append(price)
    
print(prices)
    
plt.figure(figsize=(10, 4))
plt.plot(qs, prices)
plt.xlabel('Supply')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```

```python
# Selling equation using definite integral in opposite direction

# a = max price
# b = start price
# s = supply
# q = quantity requested
# k = determines rate of gradient change

price = 0
a = 500
b = 1
k = 0
capacity = 200
q = 1
s_es = []
prices = []

s = 199
while s > 0:
    s_es.append(s)
    #price = b*q + a*math.log((capacity + k - (s - q)) / (capacity + k - s))
    price = b*q + a*math.log2((capacity + k - s + q) / (capacity + k - s))
    prices.append(price)
    s-=1
    
print(prices)
    
plt.figure(figsize=(10, 4))
plt.plot(s_es, prices)
plt.xlabel('Supply')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```

```python
# Calculating purchase prices for different supplies

# a = max price
# b = start price
# s = supply
# q = quantity requested
# k = determines rate of gradient change

price = 0
a = 500
b = 1
k = 0
capacity = 200
s = 50
qs = []
prices = []

# setting supply = 50 and quantity = 1-50
for q in range(1, 51):
    qs.append(q)
    price = b*q + a*math.log2((capacity + k - s) / (capacity + k - s - q))
    prices.append(price)

# supply = 50 and quantity = 50, do we get the same result?
price50_50 = price = b*q + a*math.log2((capacity + k - s) / (capacity + k - s - 50))

# print results
print(prices)
print()
print(price50_50)

# plot results
plt.figure(figsize=(10, 4))
plt.plot(qs, prices)
plt.xlabel('Supply')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```
