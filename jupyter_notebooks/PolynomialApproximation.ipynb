```python
# Approximation of asymptotic pricing function

# supply = token supply
# amount = requested amount of tokens

capacity = 189 # max API capacity
base_price = 100 # starting price for first token
sum_price = 0 # sum of token spot prices
spot_price = 0 # spot price for 1 token on the curve

spot_prices = [] # spot price array
sum_prices = [] # summed spot prices along the curve
supplies = [] # supply array

# point on the curve where we change from constant token price to cubic
step_point = int(capacity * 90/100)

for i in range(capacity):
    supplies.append(i)
    
# token spot prices up to the step point
for amount in range(int(step_point)):
    spot_price = base_price
    spot_prices.append(spot_price)
    sum_price += spot_price
    sum_prices.append(sum_price)
# token spot prices after the step point
for amount in range(1, capacity+1-step_point):
    spot_price = (4 * amount**3) + base_price
    spot_prices.append(spot_price)
    sum_price += spot_price
    sum_prices.append(sum_price)

#print(len(spot_prices))
#print()
print(spot_prices)
print()
print(len(sum_prices))
print()
print(sum_prices)
print()
print(sum_price)
    
plt.figure(figsize=(10, 4))
plt.plot(supplies, spot_prices)
plt.xlabel('Supply')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```

```python
# Approximation of asymptotic pricing function
# Trying to calculate/approximate summed prices without using a loop

# supply = token supply
# amount = requested amount of tokens

capacity = 100 # max API capacity
base_price = 100 # starting price for first token
sum_price = 0 # sum of token spot prices

sum_prices = [] # summed spot prices along the curve
supplies = [] # amount array

# point on the curve where we change from constant token price to cubic
step_point = int(capacity * 90/100)


s = 10
amount = 7

# loop is just to see how this works for different values of current supply
for supply in range(s, capacity):
    if supply + amount <= step_point:
        supplies.append(supply)
        # pricing function for constant portion of curve - doesn't require loop
        sum_price = base_price * amount
        sum_prices.append(sum_price)
    elif supply + amount > step_point and supply + amount < capacity:
        supplies.append(supply)
        # pricing function for cubic portion - doesn't require loop
        # x**4 is integral of 4*x**3 - the stuff in brackets is an approximation
        sum_price = (supply + amount - step_point)**4 + (base_price * amount)
        sum_prices.append(sum_price)
    else:
        break


print(len(sum_prices))
print()
print(sum_prices)
print()
print(sum_price)
    
plt.figure(figsize=(10, 4))
plt.plot(supplies, sum_prices)
plt.xlabel('Supply')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```
